```python


import os
import glob
import tkinter as tk
from tkinter import ttk, filedialog, messagebox

import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestRegressor

# Optional Word2Vec
try:
    from gensim.models import Word2Vec
    GENSIM_AVAILABLE = True
except ImportError:
    GENSIM_AVAILABLE = False


# ============================================================
# CONFIGURATION
# ============================================================

APP_NAME = "AI Energy Consumption Optimization"

ELECTRICITY_RATE = 7.5

PREFERRED_APPLIANCES = [
    "Fan",
    "Refrigerator",
    "AirConditioner",
    "Television",
    "Monitor"
]

df = pd.DataFrame()
available_appliances = []

models = {}
feature_columns = []

current_dataset = None

word2vec_model = None
tfidf_model = None
scaler = None


# ============================================================
# GUI COLORS
# ============================================================

BG_COLOR = "#F3F6FA"
SIDEBAR_COLOR = "#172033"
CARD_COLOR = "#FFFFFF"

BLUE = "#2563EB"
GREEN = "#16A34A"
RED = "#DC2626"
ORANGE = "#EA580C"
PURPLE = "#9333EA"

TEXT_COLOR = "#1E293B"
SUBTEXT_COLOR = "#64748B"


# ============================================================
# FIND DATASET AUTOMATICALLY
# ============================================================

def find_dataset():

    patterns = [
        "*.csv",
        "*.csv.zip",
        "*.zip"
    ]

    files = []

    for pattern in patterns:
        files.extend(glob.glob(pattern))

    if files:
        return files[0]

    return None


# ============================================================
# FIND COLUMN IGNORING CASE
# ============================================================

def find_column(possible_names):

    lower_columns = {
        str(column).lower().replace(" ", "").replace("_", ""): column
        for column in df.columns
    }

    for name in possible_names:

        cleaned_name = (
            str(name)
            .lower()
            .replace(" ", "")
            .replace("_", "")
        )

        if cleaned_name in lower_columns:
            return lower_columns[cleaned_name]

    return None


# ============================================================
# DETECT APPLIANCE COLUMNS
# ============================================================

def detect_appliances():

    global available_appliances

    available_appliances = []

    appliance_aliases = {

        "Fan": [
            "fan",
            "fans"
        ],

        "Refrigerator": [
            "refrigerator",
            "fridge"
        ],

        "AirConditioner": [
            "airconditioner",
            "air conditioner",
            "ac",
            "air_conditioner"
        ],

        "Television": [
            "television",
            "tv"
        ],

        "Monitor": [
            "monitor",
            "computer monitor"
        ]
    }

    renamed_columns = {}

    for standard_name, aliases in appliance_aliases.items():

        column = find_column(aliases)

        if column is not None:

            renamed_columns[column] = standard_name

    if renamed_columns:

        df.rename(
            columns=renamed_columns,
            inplace=True
        )

    for appliance in PREFERRED_APPLIANCES:

        if appliance in df.columns:

            df[appliance] = pd.to_numeric(
                df[appliance],
                errors="coerce"
            ).fillna(0)

            available_appliances.append(
                appliance
            )


# ============================================================
# LOAD DATASET
# ============================================================

def load_dataset(file_path):

    global df
    global current_dataset

    try:

        df = pd.read_csv(
            file_path,
            compression="infer"
        )

    except Exception as error:

        messagebox.showerror(
            "Dataset Error",
            f"Could not load dataset.\n\n{error}"
        )

        return False

    if df.empty:

        messagebox.showerror(
            "Dataset Error",
            "The dataset is empty."
        )

        return False

    # Clean column names
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )

    # Remove MotorPump
    motor_pump_column = find_column(
        ["MotorPump", "Motor Pump"]
    )

    if motor_pump_column is not None:

        df.drop(
            columns=[motor_pump_column],
            inplace=True
        )

    # Detect City
    city_column = find_column(
        [
            "City",
            "Location",
            "Town"
        ]
    )

    if city_column is not None and city_column != "City":

        df.rename(
            columns={city_column: "City"},
            inplace=True
        )

    if "City" not in df.columns:

        df["City"] = "Unknown"

    df["City"] = (
        df["City"]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
    )

    # Detect Company
    company_column = find_column(
        [
            "Company",
            "Provider",
            "ElectricityCompany",
            "Electricity Company"
        ]
    )

    if company_column is not None and company_column != "Company":

        df.rename(
            columns={company_column: "Company"},
            inplace=True
        )

    if "Company" not in df.columns:

        df["Company"] = "Unknown"

    df["Company"] = (
        df["Company"]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
    )

    # Detect appliance columns
    detect_appliances()

    if not available_appliances:

        messagebox.showwarning(
            "Warning",
            "No standard appliance columns were found.\n\n"
            "Expected columns such as:\n"
            "Fan, Refrigerator, AirConditioner, Television, Monitor"
        )

    # Total energy
    if available_appliances:

        df["TotalEnergy"] = (
            df[available_appliances]
            .sum(axis=1)
        )

    else:

        df["TotalEnergy"] = 0

    current_dataset = file_path

    # Prepare previous features
    prepare_minmax_scaling()
    prepare_tfidf()
    prepare_word2vec()

    # Train AI models
    train_models()

    return True


# ============================================================
# MINMAX SCALING
# ============================================================

def prepare_minmax_scaling():

    global scaler

    numeric_columns = (
        df
        .select_dtypes(include="number")
        .columns
        .tolist()
    )

    if not numeric_columns:
        return

    scaler = MinMaxScaler()

    scaled_values = scaler.fit_transform(
        df[numeric_columns]
    )

    scaled_df = pd.DataFrame(
        scaled_values,
        columns=[
            f"{column}_Scaled"
            for column in numeric_columns
        ]
    )

    # Add scaled columns
    for column in scaled_df.columns:

        df[column] = scaled_df[column]


# ============================================================
# TF-IDF FOR COMPANY
# ============================================================

def prepare_tfidf():

    global tfidf_model

    try:

        tfidf_model = TfidfVectorizer()

        tfidf_model.fit_transform(
            df["Company"]
            .fillna("")
            .astype(str)
        )

    except Exception as error:

        print(
            "TF-IDF Error:",
            error
        )


# ============================================================
# WORD2VEC FOR CITY + COMPANY
# ============================================================

def prepare_word2vec():

    global word2vec_model

    if not GENSIM_AVAILABLE:

        return

    try:

        sentences = (
            df[["City", "Company"]]
            .fillna("")
            .astype(str)
            .agg(" ".join, axis=1)
            .str.lower()
            .str.split()
            .tolist()
        )

        if sentences:

            word2vec_model = Word2Vec(
                sentences=sentences,
                vector_size=50,
                window=5,
                min_count=1,
                workers=1
            )

    except Exception as error:

        print(
            "Word2Vec Error:",
            error
        )


# ============================================================
# PREPARE FEATURES
# ============================================================

def create_features(data):

    feature_data = pd.DataFrame()

    # Encode City
    cities = sorted(
        df["City"]
        .unique()
        .tolist()
    )

    city_mapping = {
        city: index
        for index, city in enumerate(cities)
    }

    feature_data["CityCode"] = (
        data["City"]
        .map(city_mapping)
        .fillna(-1)
    )

    # Encode Company
    companies = sorted(
        df["Company"]
        .unique()
        .tolist()
    )

    company_mapping = {
        company: index
        for index, company in enumerate(companies)
    }

    feature_data["CompanyCode"] = (
        data["Company"]
        .map(company_mapping)
        .fillna(-1)
    )

    return feature_data


# ============================================================
# TRAIN RANDOM FOREST MODELS
# ============================================================

def train_models():

    global models
    global feature_columns

    models = {}

    if df.empty:

        return

    if not available_appliances:

        return

    try:

        X = create_features(df)

        feature_columns = X.columns.tolist()

        for appliance in available_appliances:

            y = df[appliance]

            model = RandomForestRegressor(
                n_estimators=100,
                random_state=42
            )

            model.fit(
                X,
                y
            )

            models[appliance] = model

    except Exception as error:

        print(
            "Model Training Error:",
            error
        )


# ============================================================
# GET CITIES
# ============================================================

def get_cities():

    if df.empty:
        return []

    return sorted(
        df["City"]
        .dropna()
        .unique()
        .tolist()
    )


# ============================================================
# GET COMPANIES FOR CITY
# ============================================================

def get_companies(city):

    if df.empty:
        return []

    filtered = df[
        df["City"] == city
    ]

    if filtered.empty:

        filtered = df

    return sorted(
        filtered["Company"]
        .dropna()
        .unique()
        .tolist()
    )


# ============================================================
# PREDICT ENERGY
# ============================================================

def predict_energy(city, company):

    predictions = {}

    if df.empty:

        return predictions, "No data"

    # Create prediction input
    input_data = pd.DataFrame(
        {
            "City": [city],
            "Company": [company]
        }
    )

    X_input = create_features(
        input_data
    )

    # Use Random Forest models
    if models:

        for appliance in available_appliances:

            if appliance in models:

                value = models[appliance].predict(
                    X_input
                )[0]

                predictions[appliance] = max(
                    0,
                    round(float(value), 2)
                )

        if predictions:

            return predictions, "AI Random Forest Prediction"

    # Fallback: City + Company average
    filtered = df[
        (df["City"] == city) &
        (df["Company"] == company)
    ]

    source = "City + Company Historical Average"

    if filtered.empty:

        filtered = df[
            df["City"] == city
        ]

        source = "City Historical Average"

    if filtered.empty:

        filtered = df[
            df["Company"] == company
        ]

        source = "Company Historical Average"

    if filtered.empty:

        filtered = df

        source = "Overall Dataset Average"

    for appliance in available_appliances:

        predictions[appliance] = round(
            float(filtered[appliance].mean()),
            2
        )

    return predictions, source


# ============================================================
# MAIN WINDOW
# ============================================================

root = tk.Tk()

root.title(APP_NAME)

root.geometry("1450x900")

root.minsize(
    1100,
    700
)

root.configure(
    bg=BG_COLOR
)


# ============================================================
# SIDEBAR
# ============================================================

sidebar = tk.Frame(
    root,
    bg=SIDEBAR_COLOR,
    width=240
)

sidebar.pack(
    side="left",
    fill="y"
)

sidebar.pack_propagate(False)


# ============================================================
# MAIN AREA
# ============================================================

main_frame = tk.Frame(
    root,
    bg=BG_COLOR
)

main_frame.pack(
    side="right",
    fill="both",
    expand=True
)


# ============================================================
# CLEAR PAGE
# ============================================================

def clear_page():

    for widget in main_frame.winfo_children():

        widget.destroy()


# ============================================================
# PAGE TITLE
# ============================================================

def create_title(title, subtitle=""):

    tk.Label(
        main_frame,
        text=title,
        bg=BG_COLOR,
        fg=TEXT_COLOR,
        font=(
            "Arial",
            24,
            "bold"
        )
    ).pack(
        anchor="w",
        padx=30,
        pady=(25, 3)
    )

    if subtitle:

        tk.Label(
            main_frame,
            text=subtitle,
            bg=BG_COLOR,
            fg=SUBTEXT_COLOR,
            font=(
                "Arial",
                11
            )
        ).pack(
            anchor="w",
            padx=30,
            pady=(0, 20)
        )


# ============================================================
# CREATE DASHBOARD CARD
# ============================================================

def create_card(parent, title, value, color):

    card = tk.Frame(
        parent,
        bg=CARD_COLOR,
        height=120
    )

    card.pack(
        side="left",
        fill="both",
        expand=True,
        padx=8
    )

    card.pack_propagate(False)

    tk.Label(
        card,
        text=title,
        bg=CARD_COLOR,
        fg=SUBTEXT_COLOR,
        font=(
            "Arial",
            11
        )
    ).pack(
        pady=(22, 5)
    )

    tk.Label(
        card,
        text=value,
        bg=CARD_COLOR,
        fg=color,
        font=(
            "Arial",
            17,
            "bold"
        )
    ).pack()


# ============================================================
# DASHBOARD
# ============================================================

def show_dashboard():

    clear_page()

    create_title(
        "Energy Dashboard",
        "AI Based Energy Consumption Optimization System"
    )

    if df.empty:

        tk.Label(
            main_frame,
            text=(
                "No dataset loaded.\n\n"
                "Open Settings and load your dataset."
            ),
            bg=BG_COLOR,
            fg=RED,
            font=(
                "Arial",
                16
            )
        ).pack(
            pady=120
        )

        return

    total_energy = float(
        df["TotalEnergy"].sum()
    )

    average_energy = float(
        df["TotalEnergy"].mean()
    )

    estimated_cost = (
        total_energy *
        ELECTRICITY_RATE
    )

    cards_frame = tk.Frame(
        main_frame,
        bg=BG_COLOR
    )

    cards_frame.pack(
        fill="x",
        padx=30
    )

    create_card(
        cards_frame,
        "Total Energy",
        f"{total_energy:.2f}",
        BLUE
    )

    create_card(
        cards_frame,
        "Average Energy",
        f"{average_energy:.2f}",
        GREEN
    )

    create_card(
        cards_frame,
        "Estimated Cost",
        f"₹ {estimated_cost:.2f}",
        PURPLE
    )

    create_card(
        cards_frame,
        "Total Records",
        str(len(df)),
        ORANGE
    )

    if available_appliances:

        graph_frame = tk.Frame(
            main_frame,
            bg=CARD_COLOR
        )

        graph_frame.pack(
            fill="both",
            expand=True,
            padx=30,
            pady=25
        )

        averages = (
            df[available_appliances]
            .mean()
        )

        figure, axis = plt.subplots(
            figsize=(10, 5)
        )

        axis.bar(
            averages.index,
            averages.values
        )

        axis.set_title(
            "Average Appliance Energy Consumption"
        )

        axis.set_xlabel(
            "Appliance"
        )

        axis.set_ylabel(
            "Average Consumption"
        )

        axis.tick_params(
            axis="x",
            rotation=20
        )

        figure.tight_layout()

        graph = FigureCanvasTkAgg(
            figure,
            master=graph_frame
        )

        graph.draw()

        graph.get_tk_widget().pack(
            fill="both",
            expand=True
        )


# ============================================================
# PREDICTION BOARD
# ============================================================

def show_prediction():

    clear_page()

    create_title(
        "AI Prediction Board",
        "Predict appliance energy consumption using City and Company"
    )

    if df.empty:

        tk.Label(
            main_frame,
            text="Please load a dataset first.",
            bg=BG_COLOR,
            fg=RED,
            font=("Arial", 15)
        ).pack(
            pady=100
        )

        return

    # INPUT CARD
    input_frame = tk.Frame(
        main_frame,
        bg=CARD_COLOR
    )

    input_frame.pack(
        fill="x",
        padx=30,
        pady=10
    )

    # CITY
    tk.Label(
        input_frame,
        text="Select City",
        bg=CARD_COLOR,
        fg=TEXT_COLOR,
        font=("Arial", 11, "bold")
    ).grid(
        row=0,
        column=0,
        padx=20,
        pady=(20, 5),
        sticky="w"
    )

    city_var = tk.StringVar()

    city_combo = ttk.Combobox(
        input_frame,
        textvariable=city_var,
        state="readonly",
        width=28
    )

    city_combo.grid(
        row=1,
        column=0,
        padx=20,
        pady=(0, 20)
    )

    # COMPANY
    tk.Label(
        input_frame,
        text="Select Company",
        bg=CARD_COLOR,
        fg=TEXT_COLOR,
        font=("Arial", 11, "bold")
    ).grid(
        row=0,
        column=1,
        padx=20,
        pady=(20, 5),
        sticky="w"
    )

    company_var = tk.StringVar()

    company_combo = ttk.Combobox(
        input_frame,
        textvariable=company_var,
        state="readonly",
        width=28
    )

    company_combo.grid(
        row=1,
        column=1,
        padx=20,
        pady=(0, 20)
    )

    # LOAD CITIES
    cities = get_cities()

    city_combo["values"] = cities

    if cities:
        city_combo.current(0)

    # UPDATE COMPANIES
    def update_companies(event=None):

        selected_city = city_var.get()

        companies = get_companies(
            selected_city
        )

        company_combo["values"] = companies

        if companies:
            company_combo.current(0)

    city_combo.bind(
        "<<ComboboxSelected>>",
        update_companies
    )

    update_companies()

    # PREDICTION RESULT CARD
    result_frame = tk.Frame(
        main_frame,
        bg=CARD_COLOR
    )

    result_frame.pack(
        fill="both",
        expand=True,
        padx=30,
        pady=15
    )

    summary_frame = tk.Frame(
        result_frame,
        bg=CARD_COLOR
    )

    summary_frame.pack(
        fill="x",
        padx=20,
        pady=(20, 5)
    )

    total_label = tk.Label(
        summary_frame,
        text="Total Predicted Energy: --",
        bg=CARD_COLOR,
        fg=BLUE,
        font=("Arial", 14, "bold")
    )

    total_label.pack(
        side="left"
    )

    cost_label = tk.Label(
        summary_frame,
        text="Predicted Cost: ₹ --",
        bg=CARD_COLOR,
        fg=GREEN,
        font=("Arial", 14, "bold")
    )

    cost_label.pack(
        side="right"
    )

    source_label = tk.Label(
        result_frame,
        text="Prediction Source: --",
        bg=CARD_COLOR,
        fg=SUBTEXT_COLOR,
        font=("Arial", 10)
    )

    source_label.pack(
        anchor="w",
        padx=20,
        pady=(5, 10)
    )

    # TABLE
    columns = (
        "Appliance",
        "Predicted",
        "Overall Average",
        "Status",
        "Recommendation"
    )

    table = ttk.Treeview(
        result_frame,
        columns=columns,
        show="headings",
        height=10
    )

    widths = {
        "Appliance": 180,
        "Predicted": 150,
        "Overall Average": 170,
        "Status": 130,
        "Recommendation": 250
    }

    for column in columns:

        table.heading(
            column,
            text=column
        )

        table.column(
            column,
            width=widths[column],
            anchor="center"
        )

    table.pack(
        fill="both",
        expand=True,
        padx=20,
        pady=15
    )

    # RUN PREDICTION
    def run_prediction():

        city = city_var.get()
        company = company_var.get()

        if not city or not company:

            messagebox.showwarning(
                "Selection Required",
                "Please select both City and Company."
            )

            return

        predictions, source = predict_energy(
            city,
            company
        )

        # Clear table
        for item in table.get_children():

            table.delete(item)

        total_prediction = 0

        for appliance in available_appliances:

            predicted = predictions.get(
                appliance,
                0
            )

            total_prediction += predicted

            overall_average = float(
                df[appliance].mean()
            )

            if predicted > overall_average * 1.10:

                status = "HIGH"

                recommendation = (
                    "Reduce usage and check efficiency"
                )

            elif predicted < overall_average * 0.90:

                status = "EFFICIENT"

                recommendation = (
                    "Good energy performance"
                )

            else:

                status = "NORMAL"

                recommendation = (
                    "Maintain current usage"
                )

            table.insert(
                "",
                "end",
                values=(
                    appliance,
                    f"{predicted:.2f}",
                    f"{overall_average:.2f}",
                    status,
                    recommendation
                )
            )

        predicted_cost = (
            total_prediction *
            ELECTRICITY_RATE
        )

        total_label.config(
            text=(
                f"Total Predicted Energy: "
                f"{total_prediction:.2f}"
            )
        )

        cost_label.config(
            text=(
                f"Predicted Cost: "
                f"₹ {predicted_cost:.2f}"
            )
        )

        source_label.config(
            text=f"Prediction Source: {source}"
        )

    predict_button = tk.Button(
        input_frame,
        text="PREDICT ENERGY",
        command=run_prediction,
        bg=GREEN,
        fg="white",
        font=("Arial", 11, "bold"),
        relief="flat",
        padx=25,
        pady=10
    )

    predict_button.grid(
        row=1,
        column=2,
        padx=25
    )


# ============================================================
# APPLIANCE GRAPHS
# ============================================================

def show_appliance_graphs():

    clear_page()

    create_title(
        "Appliance Energy Graphs",
        "Individual consumption graph for every appliance"
    )

    if df.empty or not available_appliances:

        tk.Label(
            main_frame,
            text="No appliance data available.",
            bg=BG_COLOR,
            fg=RED,
            font=("Arial", 15)
        ).pack(
            pady=100
        )

        return

    # SCROLLABLE AREA
    canvas = tk.Canvas(
        main_frame,
        bg=BG_COLOR,
        highlightthickness=0
    )

    scrollbar = ttk.Scrollbar(
        main_frame,
        orient="vertical",
        command=canvas.yview
    )

    content_frame = tk.Frame(
        canvas,
        bg=BG_COLOR
    )

    content_frame.bind(
        "<Configure>",
        lambda event:
        canvas.configure(
            scrollregion=canvas.bbox("all")
        )
    )

    canvas.create_window(
        (0, 0),
        window=content_frame,
        anchor="nw"
    )

    canvas.configure(
        yscrollcommand=scrollbar.set
    )

    canvas.pack(
        side="left",
        fill="both",
        expand=True
    )

    scrollbar.pack(
        side="right",
        fill="y"
    )

    # CREATE GRAPH FOR EVERY APPLIANCE
    for appliance in available_appliances:

        graph_card = tk.Frame(
            content_frame,
            bg=CARD_COLOR
        )

        graph_card.pack(
            fill="x",
            padx=30,
            pady=15
        )

        tk.Label(
            graph_card,
            text=f"{appliance} Energy Consumption",
            bg=CARD_COLOR,
            fg=TEXT_COLOR,
            font=("Arial", 15, "bold")
        ).pack(
            pady=(15, 5)
        )

        figure, axis = plt.subplots(
            figsize=(10, 4)
        )

        axis.plot(
            range(1, len(df) + 1),
            df[appliance].values,
            marker="o"
        )

        axis.set_title(
            f"{appliance} Consumption Trend"
        )

        axis.set_xlabel(
            "Record Number"
        )

        axis.set_ylabel(
            "Energy Consumption"
        )

        axis.grid(
            True,
            alpha=0.3
        )

        figure.tight_layout()

        graph = FigureCanvasTkAgg(
            figure,
            master=graph_card
        )

        graph.draw()

        graph.get_tk_widget().pack(
            fill="both",
            expand=True,
            padx=20,
            pady=15
        )


# ============================================================
# DATA TABLE PAGE
# ============================================================

def show_data():

    clear_page()

    create_title(
        "Dataset Explorer",
        "View the loaded electricity consumption dataset"
    )

    if df.empty:

        tk.Label(
            main_frame,
            text="No dataset loaded.",
            bg=BG_COLOR,
            fg=RED,
            font=("Arial", 15)
        ).pack(
            pady=100
        )

        return

    table_frame = tk.Frame(
        main_frame,
        bg=CARD_COLOR
    )

    table_frame.pack(
        fill="both",
        expand=True,
        padx=30,
        pady=15
    )

    columns = list(df.columns)

    tree = ttk.Treeview(
        table_frame,
        columns=columns,
        show="headings"
    )

    for column in columns:

        tree.heading(
            column,
            text=column
        )

        tree.column(
            column,
            width=130,
            anchor="center"
        )

    for _, row in df.head(100).iterrows():

        tree.insert(
            "",
            "end",
            values=[
                str(value)
                for value in row.tolist()
            ]
        )

    horizontal = ttk.Scrollbar(
        table_frame,
        orient="horizontal",
        command=tree.xview
    )

    vertical = ttk.Scrollbar(
        table_frame,
        orient="vertical",
        command=tree.yview
    )

    tree.configure(
        xscrollcommand=horizontal.set,
        yscrollcommand=vertical.set
    )

    tree.pack(
        side="left",
        fill="both",
        expand=True
    )

    vertical.pack(
        side="right",
        fill="y"
    )

    horizontal.pack(
        side="bottom",
        fill="x"
    )


# ============================================================
# SETTINGS / DATASET LOADER
# ============================================================

def show_settings():

    clear_page()

    create_title(
        "Settings",
        "Load and manage your electricity dataset"
    )

    card = tk.Frame(
        main_frame,
        bg=CARD_COLOR
    )

    card.pack(
        fill="x",
        padx=30,
        pady=20
    )

    dataset_name = (
        os.path.basename(current_dataset)
        if current_dataset
        else "No dataset loaded"
    )

    details = [
        ("Dataset", dataset_name),
        ("Records", len(df)),
        (
            "Appliances",
            ", ".join(available_appliances)
            if available_appliances
            else "Not detected"
        ),
        ("AI Models", len(models)),
        (
            "Word2Vec",
            "Enabled"
            if word2vec_model is not None
            else "Optional / Not Available"
        ),
        (
            "TF-IDF",
            "Enabled"
            if tfidf_model is not None
            else "Not Available"
        )
    ]

    for title, value in details:

        row = tk.Frame(
            card,
            bg=CARD_COLOR
        )

        row.pack(
            fill="x",
            padx=25,
            pady=12
        )

        tk.Label(
            row,
            text=title,
            bg=CARD_COLOR,
            fg=TEXT_COLOR,
            font=("Arial", 11, "bold")
        ).pack(
            side="left"
        )

        tk.Label(
            row,
            text=str(value),
            bg=CARD_COLOR,
            fg=SUBTEXT_COLOR,
            font=("Arial", 10)
        ).pack(
            side="right"
        )

    # LOAD FILE FUNCTION
    def choose_dataset():

        file_path = filedialog.askopenfilename(
            title="Select Electricity Dataset",
            filetypes=[
                ("CSV Files", "*.csv"),
                ("ZIP Files", "*.zip"),
                ("All Files", "*.*")
            ]
        )

        if not file_path:
            return

        success = load_dataset(
            file_path
        )

        if success:

            messagebox.showinfo(
                "Success",
                "Dataset loaded and AI models trained successfully."
            )

            show_dashboard()

    tk.Button(
        main_frame,
        text="LOAD DATASET",
        command=choose_dataset,
        bg=BLUE,
        fg="white",
        font=("Arial", 11, "bold"),
        relief="flat",
        padx=30,
        pady=12
    ).pack(
        pady=10
    )


# ============================================================
# NAVIGATION BUTTON
# ============================================================

def create_nav_button(text, command):

    tk.Button(
        sidebar,
        text=text,
        command=command,
        bg=SIDEBAR_COLOR,
        fg="white",
        activebackground="#29364E",
        activeforeground="white",
        relief="flat",
        anchor="w",
        padx=25,
        pady=13,
        font=("Arial", 11)
    ).pack(
        fill="x",
        padx=10,
        pady=3
    )


# ============================================================
# SIDEBAR HEADER
# ============================================================

tk.Label(
    sidebar,
    text="⚡ AI ENERGY",
    bg=SIDEBAR_COLOR,
    fg="white",
    font=("Arial", 19, "bold")
).pack(
    pady=(35, 5)
)

tk.Label(
    sidebar,
    text="Optimization System",
    bg=SIDEBAR_COLOR,
    fg="#94A3B8",
    font=("Arial", 10)
).pack(
    pady=(0, 35)
)


# ============================================================
# SIDEBAR MENU
# ============================================================

create_nav_button(
    "🏠  Dashboard",
    show_dashboard
)

create_nav_button(
    "🤖  AI Prediction Board",
    show_prediction
)

create_nav_button(
    "📊  Appliance Graphs",
    show_appliance_graphs
)

create_nav_button(
    "📋  Dataset Explorer",
    show_data
)

create_nav_button(
    "⚙  Settings",
    show_settings
)


# ============================================================
# FOOTER
# ============================================================

tk.Label(
    sidebar,
    text="AI Based Energy Optimization",
    bg=SIDEBAR_COLOR,
    fg="#64748B",
    font=("Arial", 8)
).pack(
    side="bottom",
    pady=20
)


# ============================================================
# LOAD DEFAULT DATASET
# ============================================================

default_dataset = find_dataset()

if default_dataset:

    try:

        load_dataset(
            default_dataset
        )

        print(
            "Dataset loaded:",
            default_dataset
        )

    except Exception as error:

        print(
            "Automatic dataset loading error:",
            error
        )


# ============================================================
# START APPLICATION
# ============================================================

show_dashboard()

root.mainloop()
```
